```python
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("All packages are working!")
```

```python
# 1. Read / Acquire Image
image = cv2.imread("tree.jpg")
```

```python
if image is None:
    print("Image could not be loaded.")
else:
    print("Image loaded successfully.")
```

```python
print("Image shape:", image.shape)
```

```python
   # 2. Display Image Information
height, width, channels = image.shape

print("Height:", height)
print("Width:", width)
print("Channels:", channels)
print("Total pixels:", height * width)
```

```python
 # 3. Convert BGR to RGB for Matplotlib
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
# 4. Save Original Image
cv2.imwrite("output_original.jpg", image)
```

```python
# 5. Grayscale Conversion
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

```python
cv2.imwrite("grayscale.jpg", gray)
```

```python
   # 6. Resize
resized = cv2.resize(image, (400, 300))
```

```python
resized_rgb = cv2.cvtColor(resized, cv2.COLOR_BGR2RGB)

plt.imshow(resized_rgb)
plt.title("Resized Image")
plt.axis("off")
plt.show()
```

```python
cv2.imwrite("resized.jpg", resized)
```

```python
 #7. Rotation
rotated = cv2.rotate(image, cv2.ROTATE_90_CLOCKWISE)
```

```python
rotated_rgb = cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB)

plt.imshow(rotated_rgb)
plt.title("Rotated Image")
plt.axis("off")
plt.show()
```

```python
 # 8. Horizontal Flip
horizontal_flip = cv2.flip(image, 1)
```

```python
horizontal_rgb = cv2.cvtColor(horizontal_flip, cv2.COLOR_BGR2RGB)

plt.imshow(horizontal_rgb)
plt.title("Horizontal Flip")
plt.axis("off")
plt.show()
```

```python
# 9. Vertical Flip
vertical_flip = cv2.flip(image, 0)
```

```python
vertical_rgb = cv2.cvtColor(vertical_flip, cv2.COLOR_BGR2RGB)

plt.imshow(vertical_rgb)
plt.title("Vertical Flip")
plt.axis("off")
plt.show()
```

```python
# 10. Negative / Complement
negative = 255 - image
```

```python
negative_rgb = cv2.cvtColor(negative, cv2.COLOR_BGR2RGB)

plt.imshow(negative_rgb)
plt.title("Negative Image")
plt.axis("off")
plt.show()
```

```python
cv2.imwrite("negative.jpg", negative)
```

```python
 # 11. Crop ROI
roi = image[235:460, 460:735]
```

```python
roi_rgb = cv2.cvtColor(roi, cv2.COLOR_BGR2RGB)

plt.imshow(roi_rgb)
plt.title("Region of Interest")
plt.axis("off")
plt.show()
```

```python
cv2.imwrite("roi.jpg", roi)
```

```python
print("ROI shape:", roi.shape)
```

```python
 # 12. Analyze ROI
roi_gray = cv2.cvtColor(roi, cv2.COLOR_BGR2GRAY)

print("ROI mean intensity:", np.mean(roi_gray))
```

```python
print("ROI minimum intensity:", np.min(roi_gray))
print("ROI maximum intensity:", np.max(roi_gray))
```

```python
    # 13. Display All Images

plt.figure(figsize=(12, 8))

plt.subplot(3, 3, 1)
plt.imshow(image_rgb)
plt.title("Original")
plt.axis("off")

plt.subplot(3, 3, 2)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale")
plt.axis("off")

plt.subplot(3, 3, 3)
plt.imshow(resized_rgb)
plt.title("Resized")
plt.axis("off")

plt.subplot(3, 3, 4)
plt.imshow(rotated_rgb)
plt.title("Rotated")
plt.axis("off")
plt.subplot(3, 3, 5)
plt.imshow(horizontal_rgb)
plt.title("Horizontal Flip")
plt.axis("off")

plt.subplot(3, 3, 6)
plt.imshow(vertical_rgb)
plt.title("Vertical Flip")
plt.axis("off")

plt.subplot(3, 3, 7)
plt.imshow(negative_rgb)
plt.title("Negative")
plt.axis("off")

plt.subplot(3, 3, 8)
plt.imshow(roi_rgb)
plt.title("ROI")
plt.axis("off")

plt.tight_layout()
plt.show()
```
