```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

print("All libraries installed successfully!")
```

```python
img = cv2.imread('img.jpg')
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
plt.figure(figsize=(8, 6))
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
cv2.imwrite("original_image.jpg", cv2.cvtColor(img_rgb, cv2.COLOR_RGB2BGR))
```

```python
print("Image shape:", img.shape)
print("Image height:", img.shape[0])
print("Image width:", img.shape[1])
print("Number of channels:", img.shape[2])
```

```python
gaussian = cv2.GaussianBlur(img_rgb, (5, 5), 0)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(gaussian)
plt.title("Gaussian Blur")
plt.axis("off")

plt.show()
```

```python

cv2.imwrite("GussianBlur_image.jpg", cv2.cvtColor(img_rgb, cv2.COLOR_RGB2BGR))
```

```python
median = cv2.medianBlur(img_rgb, 5)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(median)
plt.title("Median Filter")
plt.axis("off")

plt.show()
```

```python

cv2.imwrite("MedianFilter_image.jpg", cv2.cvtColor(img_rgb, cv2.COLOR_RGB2BGR))
```

```python
average = cv2.blur(img_rgb, (5, 5))
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(average)
plt.title("Average / Mean Filter")
plt.axis("off")

plt.show()
```

```python

cv2.imwrite("Average_image.jpg", cv2.cvtColor(img_rgb, cv2.COLOR_RGB2BGR))
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 4, 1)
plt.imshow(img_rgb)
plt.title("Original")
plt.axis("off")

plt.subplot(1, 4, 2)
plt.imshow(average)
plt.title("Average Filter")
plt.axis("off")

plt.subplot(1, 4, 3)
plt.imshow(gaussian)
plt.title("Gaussian Filter")
plt.axis("off")

plt.subplot(1, 4, 4)
plt.imshow(median)
plt.title("Median Filter")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)
```

```python
plt.imshow(gray, cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

```python

cv2.imwrite("grayScale_image.jpg", cv2.cvtColor(gray, cv2.COLOR_RGB2BGR))
```

```python
laplacian = cv2.Laplacian(gray, cv2.CV_64F)
laplacian = np.absolute(laplacian)
laplacian = np.uint8(laplacian)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(gray, cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(laplacian, cmap='gray')
plt.title("Laplacian Filter")
plt.axis("off")

plt.show()
```

```python

cv2.imwrite("Laplacian_image.jpg", cv2.cvtColor(laplacian, cv2.COLOR_RGB2BGR))
```

```python
sobel_x = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
sobel_x = np.absolute(sobel_x)
sobel_x = np.uint8(sobel_x)
```

```python

cv2.imwrite("sobelX_image.jpg", cv2.cvtColor(sobel_x, cv2.COLOR_RGB2BGR))
```

```python
sobel_y = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)
sobel_y = np.absolute(sobel_y)
sobel_y = np.uint8(sobel_y)
```

```python

cv2.imwrite("sobelY_image.jpg", cv2.cvtColor(gray, cv2.COLOR_RGB2BGR))
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(gray, cmap='gray')
plt.title("Original Grayscale")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(sobel_x, cmap='gray')
plt.title("Sobel X - Vertical Edges")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(sobel_y, cmap='gray')
plt.title("Sobel Y - Horizontal Edges")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
sobel_combined = cv2.magnitude(
    cv2.Sobel(gray, cv2.CV_32F, 1, 0, ksize=3),
    cv2.Sobel(gray, cv2.CV_32F, 0, 1, ksize=3)
)

sobel_combined = cv2.convertScaleAbs(sobel_combined)
```

```python
plt.imshow(sobel_combined, cmap='gray')
plt.title("Combined Sobel Edge Detection")
plt.axis("off")
plt.show()
```

```python

cv2.imwrite("combinedSobel_image.jpg", cv2.cvtColor(sobel_combined, cv2.COLOR_RGB2BGR))
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(gray, cmap='gray')
plt.title("Grayscale")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(laplacian, cmap='gray')
plt.title("Laplacian")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(sobel_combined, cmap='gray')
plt.title("Sobel")
plt.axis("off")

plt.tight_layout()
plt.show()
```
